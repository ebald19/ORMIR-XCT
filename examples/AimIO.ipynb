<a name="reading"></a>
## *Step 2: Reading:*

Import modules/packages and set the input image path. 

```python
from pathlib import Path
from py_aimio import read_aim, write_aim

# Specify the directory path
directory_path = Path("images\AIMIO_Examples")

# Loop through all test files
for file_path in directory_path.iterdir():
    if file_path.is_file():
        print(f"Reading File: {file_path.name}")

    array, meta = read_aim(str(file_path))

    # display image contents using matlab
    print(array[:10])
```

<a name="writing"></a>
## *Step 3: Writing:*

Write AIM images

```python
# Specify the directory path
test_image_dir = Path("images\AIMIO_Examples")

# Loop through all test files
for file_path in test_image_dir.iterdir():
    array, meta = read_aim(str(file_path))

    out_path = Path("outputs") / Path(file_path.name)
    print(f"Writing File: {out_path.name}")
    write_aim(str(out_path), array, meta)


# at this point, all images should be reproduced in the outputs folder
```
